### Тапсырма 22. Үшбұрыштың түрі

Үшбұрыштың үш қабырғасының ұзындығы `a`, `b`, `c` берілген. Алдымен мұндай үшбұрыштың болуы мүмкін екенін тексеріңіз: кез келген екі қабырғаның қосындысы үшінші қабырғадан үлкен болуы керек. Үшбұрыш бар болса, оның тең қабырғалы, теңбүйірлі немесе әртүрлі қабырғалы екенін анықтаңыз.

````python
a = 5
b = 5
c = 5

if a + b > c and a + c > b and b + c > a:
    if a == b and b == c:
        print("Тең қабырғалы үшбұрыш")
    elif a == b or a == c or b == c:
        print("Теңбүйірлі үшбұрыш")
    else:
        print("Әртүрлі қабырғалы үшбұрыш")
else:
    print("Мұндай үшбұрыш жоқ")
````

### Тапсырма 23. Қарапайым калькулятор

````python
a = 15
b = 5
operation = "+"

if operation == "+":
    print(a + b)
elif operation == "-":
    print(a - b)
elif operation == "*":
    print(a * b)
elif operation == "/":
    if b != 0:
        print(a / b)
    else:
        print("Нөлге бөлуге болмайды")
else:
    print("Белгісіз амал")
````

### Тапсырма 28. Сандарды санау

`numbers = [12, -5, 8, 0, -3, 17, 4, 0, -9]` тізімі берілген. Тізімдегі оң, теріс сандардың және нөлдердің санын бөлек есептеп шығарыңыз.

**Күтілетін нәтиже:**

```
Оң сандар: 4
Теріс сандар: 3
Нөлдер: 2
```

Фибоначчи тізбегінің алғашқы `n` мүшесінен тұратын тізімді қайтаратын `fibonacci(n)` функциясын жазыңыз. Тізбектің алғашқы екі мүшесі 0 және 1, ал әрбір келесі мүше алдыңғы екі мүшенің қосындысына тең.

### Тапсырма 44. Тізімнің статистикасы

Сандар тізімін қабылдап, оның ең кіші мәнін, ең үлкен мәнін және орташа мәнін кортеж түрінде қайтаратын `stats(numbers)` функциясын жазыңыз. `min()` және `max()` функцияларын қолданбаңыз. Бос тізім берілсе, функция `None` қайтаруы керек.

**Күтілетін нәтиже:**

```
stats([4, 9, 1, 7, 4]) -> (1, 9, 5.0)
stats([]) -> None
```

````python
def stats(numbers):
    if len(numbers) == 0:
        return None

    minimum = numbers[0]
    maximum = numbers[0]
    total = 0

    for i in numbers:
        if i < minimum:
            minimum = i

        if i > maximum:
            maximum = i

        total += i

    average = total / len(numbers)

    return (minimum, maximum, average)

print(stats([4, 9, 1, 7, 4]))
print(stats([]))
````

### Тапсырма 46. Құпиясөзді тексеру

Құпиясөздің сенімділігін тексеретін `validate_password(password)` функциясын жазыңыз. Құпиясөз кемінде 8 символдан тұруы, кемінде бір цифр және кемінде бір бас әріп қамтуы керек. Функция құпиясөз сенімді болса "Құпиясөз сенімді" деп, әйтпесе қай талап орындалмағанын көрсететін хабарламаны қайтаруы тиіс. Жол әдістерінен `isdigit()` және `isupper()` пайдалы болуы мүмкін.

**Күтілетін нәтиже:**

```
validate_password("abc") -> Құпиясөз кемінде 8 символдан тұруы керек
validate_password("abcdefgh1") -> Құпиясөзде бас әріп болуы керек
validate_password("Abcdefgh1") -> Құпиясөз сенімді
```

````python
def validate_password(password):

    if len(password) < 8:
        return "Құпиясөз кемінде 8 символдан тұруы керек"

    has_digit = False
    has_upper = False

    for i in password:
        if i.isdigit():
            has_digit = True

        if i.isupper():
            has_upper = True

    if not has_digit:
        return "Құпиясөзде цифр болуы керек"

    if not has_upper:
        return "Құпиясөзде бас әріп болуы керек"

    return "Құпиясөз сенімді"

print(validate_password("abc"))
print(validate_password("abcdefgh1"))
print(validate_password("Abcdefgh1"))
````

### Тапсырма 47. Банкомат симуляциясы

Бастапқы баланс `balance = 10000`. Операциялар тізімі берілген:

`operations = [("deposit", 5000), ("withdraw", 3000), ("withdraw", 15000), ("deposit", 2000), ("withdraw", 4000)]`

Әр операцияны орындайтын `process(balance, operation)` функциясын жазыңыз. `deposit` балансты көбейтеді, `withdraw` азайтады. Қаражат жеткіліксіз болса, операция орындалмай, ескерту шығуы керек. Барлық операциядан кейінгі соңғы балансты шығарыңыз.

**Күтілетін нәтиже:**

```
Соңғы баланс: 10000
```

````python
def process(balance, operation):

    action, amount = operation

    if action == "deposit":
        balance += amount

    elif action == "withdraw":
        if balance >= amount:
            balance -= amount
        else:
            print("Қаражат жеткіліксіз")

    return balance


balance = 10000

operations = [
    ("deposit", 5000),
    ("withdraw", 3000),
    ("withdraw", 15000),
    ("deposit", 2000),
    ("withdraw", 4000)
]

for operation in operations:
    balance = process(balance, operation)

print("Соңғы баланс:", balance)
````

### Тапсырма 48. Студенттер журналы

Студенттердің баллдары сөздікте берілген:

`journal = {"Aigerim": [85, 92, 78], "Nurlan": [65, 70, 58], "Dana": [95, 98, 91], "Arman": [45, 52, 40]}`

Келесі функцияларды жазыңыз: орташа баллды есептейтін `average(scores)`, баллға сәйкес бағаны (A, B, C, F) қайтаратын `get_grade(score)`, үздік студенттің атын қайтаратын `best_student(journal)`. Осы функцияларды қолданып, әр студенттің аты, орташа баллы мен бағасы көрсетілген кестені және үздік студентті шығарыңыз.

````python
journal = {
    "Aigerim": [85, 92, 78],
    "Nurlan": [65, 70, 58],
    "Dana": [95, 98, 91],
    "Arman": [45, 52, 40]
}

def average(scores):
    return sum(scores) / len(scores)


def get_grade(score):
    if score >= 90:
        return "A"
    elif score >= 75:
        return "B"
    elif score >= 50:
        return "C"
    else:
        return "F"


def best_student(journal):
    best_name = ""
    best_score = 0

    for name, scores in journal.items():
        avg = average(scores)

        if avg > best_score:
            best_score = avg
            best_name = name

    return best_name


print("Студент | Орташа балл | Баға")

for name, scores in journal.items():
    avg = average(scores)
    grade = get_grade(avg)

    print(name, "|", round(avg, 2), "|", grade)

print("Үздік студент:", best_student(journal))
````

Болжамдарды кезекпен тексеретін бағдарлама жазыңыз: әр болжам үшін "Ойлаған сан кішірек", "Ойлаған сан үлкенірек" немесе "Таптыңыз!" деп шығарыңыз. Сан табылғанда цикл тоқтап, неше әрекетте табылғаны көрсетілуі керек. Тексеру логикасын `check_guess(secret, guess)` функциясына бөліп жазыңыз.

**Күтілетін нәтиже:**

```
50: Ойлаған сан кішірек
25: Ойлаған сан үлкенірек
37: Ойлаған сан үлкенірек
45: Ойлаған сан кішірек
42: Таптыңыз!
Әрекеттер саны: 5
```

### Тапсырма 50. Цезарь шифры

Мәтінді Цезарь шифрымен шифрлайтын `encrypt(text, shift)` функциясын жазыңыз: әрбір латын әрпі алфавит бойынша `shift` орынға жылжытылады, ал алфавит соңына жеткенде басынан жалғасады. Бас және кіші әріптердің регистрі сақталуы, ал әріп емес символдар өзгеріссіз қалуы керек. Содан кейін шифрланған мәтінді бастапқы күйіне қайтаратын `decrypt(text, shift)` функциясын жазыңыз. `ord()` және `chr()` функциялары пайдалы болуы мүмкін.

**Күтілетін нәтиже:**

```
encrypt("Hello, World!", 3) -> Khoor, Zruog!
decrypt("Khoor, Zruog!", 3) -> Hello, World!
```

````python
def encrypt(text, shift):
    result = ""

    for letter in text:
        if letter.isupper():
            new_letter = chr((ord(letter) - ord("A") + shift) % 26 + ord("A"))
            result += new_letter

        elif letter.islower():
            new_letter = chr((ord(letter) - ord("a") + shift) % 26 + ord("a"))
            result += new_letter

        else:
            result += letter

    return result


def decrypt(text, shift):
    return encrypt(text, -shift)


encrypted = encrypt("Hello, World!", 3)
decrypted = decrypt(encrypted, 3)

print(encrypted)
print(decrypted)
````
